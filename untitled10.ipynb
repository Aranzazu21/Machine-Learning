```python
import pandas as pd
import numpy as np
from google.colab import drive

#  Montar Google Drive
print("Conectando a Google Drive...")
drive.mount('/content/drive')

# Cargar tu dataset de Excel predefinido (omitiendo metadatos)
ruta_excel = '/content/drive/MyDrive/Ciencia de datos /2026-08-18-20-04-31-weight.xlsx'
df = pd.read_excel(ruta_excel, skiprows=3)


# Renombrar columnas a un formato manejable según la rúbrica
df = df.rename(columns={
    'Peso(Kg)': 'Peso_Kg',
    'Agua(%)': 'Agua_%',
    'Porcentaje de Grasa Corporal': 'Grasa_%',
    'Porcentaje de masa muscular esquelética': 'Musculo_%'
})

# Convertir la fecha a formato datetime (Agregamos '2026-' porque el Excel solo trae Mes-Día)
df['Fecha_DT'] = pd.to_datetime('2026-' + df['Fecha'].astype(str), format='%Y-%m-%d', errors='coerce')
df = df.sort_values('Fecha_DT').reset_index(drop=True)

# Forzar columnas biométricas a formato numérico (por si hay guiones '--' de errores de báscula)
cols_bio = ['Peso_Kg', 'Agua_%', 'Grasa_%', 'Musculo_%']
df[cols_bio] = df[cols_bio].apply(pd.to_numeric, errors='coerce')
df[cols_bio] = df[cols_bio].interpolate().ffill().bfill() # Llenar huecos

# Validar columnas de actividad física
for col in ['Distancia_MTB_km', 'Volumen_Ciclismo', 'Calorias_Fuerza']:
    if col not in df.columns:
        df[col] = 0.0


# Media móvil y residuos
mm_peso = df['Peso_Kg'].rolling(window=7, min_periods=1).mean()
residuo_peso = df['Peso_Kg'] - mm_peso

Q1 = residuo_peso.quantile(0.25)
Q3 = residuo_peso.quantile(0.75)
IQR = Q3 - Q1

# Filtrar: Caída abrupta menor al límite inferior Y rodada > 40km
condicion_anomalia = (residuo_peso < (Q1 - 1.5 * IQR)) & (df['Distancia_MTB_km'] > 40)

# Suavizar picos aislados imputando la media móvil
df.loc[condicion_anomalia, 'Peso_Kg'] = mm_peso


# Fin de semana
df['es_fin_semana'] = df['Fecha_DT'].dt.dayofweek.isin([5, 6]).astype(int)

#  Carga semanal en MTB Suma móvil de 7 días
df['carga_semanal_km'] = df['Distancia_MTB_km'].rolling(window=7, min_periods=1).sum()

# Ratio Fuerza / Cardio Usamos np.where para evitar errores de división por cero
df['ratio_fuerza_cardio'] = np.where(df['Volumen_Ciclismo'] > 0, df['Calorias_Fuerza'] / df['Volumen_Ciclismo'], 0)

# Distancia al normopeso (IMC Meta)
df['distancia_al_normopeso'] = df['Peso_Kg'] - 70.28


for col in cols_bio:
    df[f'{col}_Z'] = (df[col] - df[col].mean()) / df[col].std()


# Asegurar que no queden valores nulos sin tratar como lo pide la rúbrica
df = df.fillna(0)

# Exportar Dataset Limpio y Transformado (Generamos ambos formatos por seguridad)
df.to_csv('dataset_limpio_transformado.csv', index=False)
df.to_excel('dataset_limpio_transformado.xlsx', index=False)

print("Pipeline P2P2 ejecutado correctamente.")
print("Archivos exportados exitosamente sin valores nulos.")


# Mostrar en pantalla días atípicos (Z > 1.5 o Z < -1.5) para el reporte de Word
print("\n=== ATENCIÓN: DÍAS ATÍPICOS (Para tu Reporte) ===")
dias_atipicos = df[
    (df['Musculo_%_Z'] > 1.5) | (df['Musculo_%_Z'] < -1.5) |
    (df['Grasa_%_Z'] > 1.5) | (df['Grasa_%_Z'] < -1.5)
]
display(dias_atipicos[['Fecha_DT', 'Peso_Kg', 'Grasa_%_Z', 'Musculo_%_Z']].head(5))
```
