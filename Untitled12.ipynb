```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats
from google.colab import drive
import warnings
warnings.filterwarnings('ignore')

print("Conectando a Google Drive...")
drive.mount('/content/drive')

ruta_excel = '/content/drive/MyDrive/Ciencia de datos /dataset_limpio_transformado (2).xlsx'
df = pd.read_excel(ruta_excel)
df['Fecha_DT'] = pd.to_datetime(df['Fecha_DT'])

print("\nmatris de correlacion pearson y spearman")
cols_corr = ['Peso_Kg', 'Grasa_%', 'Musculo_%', 'carga_semanal_km', 'conteo_pasos']

corr_pearson = df[cols_corr].corr(method='pearson')
corr_spearman = df[cols_corr].corr(method='spearman')

sns.set_theme(style="white")
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.heatmap(corr_pearson, annot=True, cmap='coolwarm', fmt=".2f", ax=axes[0], cbar=False)
axes[0].set_title('matriz de pearson (lineal)', fontsize=14, fontweight='bold')

sns.heatmap(corr_spearman, annot=True, cmap='viridis', fmt=".2f", ax=axes[1])
axes[1].set_title('matriz de spearman (no lineal)', fontsize=14, fontweight='bold')

plt.tight_layout()
plt.show()

corr_mancuernas = df['Musculo_%'].corr(df['Calorias_Fuerza'])
corr_ciclismo = df['Musculo_%'].corr(df['Distancia_MTB_km'])

corr_mancuernas = 0 if pd.isna(corr_mancuernas) else corr_mancuernas
corr_ciclismo = 0 if pd.isna(corr_ciclismo) else corr_ciclismo

print(f"\nCorrelacion musculo vs mancuernas: {corr_mancuernas:.3f}")
print(f"Correlacion musculo vs ciclismo: {corr_ciclismo:.3f}")

if corr_mancuernas > corr_ciclismo:
    print("Respuesta: El porcentaje de musculo aumenta mas con las sesiones de mancuernas.")
elif corr_ciclismo > corr_mancuernas:
    print("Respuesta: El porcentaje de musculo aumenta mas con el kilometraje de ciclismo.")
else:
    print("Respuesta: En este corte de datos, no hay volumen estadistico suficiente para una diferencia clara.")

print("\nagrupacion por dia de la semana")
df['Dia_Semana'] = df['Fecha_DT'].dt.day_name()
dias_orden = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
df['Dia_Semana'] = pd.Categorical(df['Dia_Semana'], categories=dias_orden, ordered=True)

resumen_dias = df.groupby('Dia_Semana')[['Distancia_MTB_km', 'Calorias_Fuerza']].mean().reset_index()
resumen_melt = resumen_dias.melt(id_vars='Dia_Semana', var_name='Actividad', value_name='Promedio')

plt.figure(figsize=(10, 5))
sns.barplot(data=resumen_melt, x='Dia_Semana', y='Promedio', hue='Actividad', palette='Set2')
plt.title('estacionalidad de entrenamiento por dia', fontsize=14, fontweight='bold')
plt.ylabel('promedio (km o kcal)')
plt.xlabel('dia de la semana')
plt.legend(title='tipo de ejercicio')
plt.show()

print("\nprueba t ")
df['Semana_Año'] = df['Fecha_DT'].dt.isocalendar().week
resumen_semanal = df.groupby('Semana_Año').agg(
    Volumen_Total_km=('Distancia_MTB_km', 'sum'),
    Peso_Max=('Peso_Kg', 'max'),
    Peso_Min=('Peso_Kg', 'min')
).reset_index()

resumen_semanal['Perdida_Peso'] = resumen_semanal['Peso_Max'] - resumen_semanal['Peso_Min']

semanas_alto = resumen_semanal[resumen_semanal['Volumen_Total_km'] > 80]['Perdida_Peso']
semanas_bajo = resumen_semanal[resumen_semanal['Volumen_Total_km'] < 40]['Perdida_Peso']

print(f"Semanas de alto volumen (>80km): {len(semanas_alto)}")
print(f"Semanas de bajo volumen (<40km): {len(semanas_bajo)}")

if len(semanas_alto) > 1 and len(semanas_bajo) > 1:
    t_stat, p_value = stats.ttest_ind(semanas_alto, semanas_bajo, equal_var=False)
    print(f"\nEstadistico t: {t_stat:.4f}")
    print(f"Valor P (p-value): {p_value:.4f}")
else:
    print("\nNo hay suficientes datos (semanas >80km) en el dataset para arrojar el p-value.")
```
