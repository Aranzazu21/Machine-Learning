```python
import pandas as pd
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.metrics import r2_score
from google.colab import drive
import warnings
warnings.filterwarnings('ignore')


drive.mount('/content/drive')


ruta = '/content/drive/MyDrive/Ciencia de datos /dataset_limpio_transformado (2).xlsx'
df = pd.read_excel(ruta)
df['Fecha_DT'] = pd.to_datetime(df['Fecha_DT'])
df = df.sort_values('Fecha_DT').reset_index(drop=True)

# variable de tiempo t
df['Dias_Transcurridos'] = (df['Fecha_DT'] - df['Fecha_DT'].min()).dt.days

# variables predictoras
cols = ['Dias_Transcurridos', 'carga_semanal_km', 'Calorias_Fuerza', 'conteo_pasos', 'tiempo_actividad_min']
df[cols] = df[cols].fillna(0)
df['Peso_Kg'] = df['Peso_Kg'].fillna(df['Peso_Kg'].mean())

X = df[cols]
y = df['Peso_Kg']

# entrenar modelos
ols = LinearRegression().fit(X, y)
ridge = Ridge(alpha=1.0).fit(X, y)
lasso = Lasso(alpha=0.1).fit(X, y)

print("\n paso 1 regrsion lineal multiuple ")
print("modelo ajustado prediciendo Peso(Kg) segun la rubrica:")
print(f"Ecuacion: Y = {ols.intercept_:.4f} + ({ols.coef_[0]:.4f} * Dias) + ({ols.coef_[1]:.4f} * Carga_Km) + ({ols.coef_[2]:.4f} * Kcal_Fuerza)")

print("\npaso 2 comparacion de old ,ridge y lasso  ")
print("evaluando colinealidad entre pasos y minutos activos:")
print(f"R2 OLS:   {r2_score(y, ols.predict(X)):.4f}")
print(f"R2 Ridge: {r2_score(y, ridge.predict(X)):.4f} (Regularizacion L2)")
print(f"R2 Lasso: {r2_score(y, lasso.predict(X)):.4f} (Regularizacion L1)")

print("\npaso 3 proyeccion temporal")
meta = 70.28
min_diarios = 1226 / 30.41 # ritmo mensual a diario

# sumar el impacto historico de las otras variables para aislar t
constantes = (ols.coef_[1] * df['carga_semanal_km'].mean() +
              ols.coef_[2] * df['Calorias_Fuerza'].mean() +
              ols.coef_[3] * df['conteo_pasos'].mean() +
              ols.coef_[4] * min_diarios)

coef_tiempo = ols.coef_[0]
intercepto = ols.intercept_

if coef_tiempo >= 0:
    print("el peso subio o se estanco, el coeficiente de tiempo es positivo.")
    print("matematicamente no se puede despejar una fecha futura hacia la baja.")
else:
    print("despejando la variable temporal t de la ecuacion...")
    print("t = (Meta - B0 - Impacto_Otras_Variables) / B1")

    dias_meta = (meta - intercepto - constantes) / coef_tiempo
    dias_faltantes = dias_meta - df['Dias_Transcurridos'].max()

    if dias_faltantes > 0:
        fecha_meta = df['Fecha_DT'].max() + pd.Timedelta(days=int(dias_faltantes))
        print(f"\ndias restantes para alcanzar meta: {int(dias_faltantes)}")
        print(f"fecha estimada: {fecha_meta.strftime('%Y-%m-%d')}")
    else:
        print("\nla meta ya se alcanzo o el tiempo calculado es negativo")
```
