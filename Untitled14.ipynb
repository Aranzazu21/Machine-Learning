```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import drive
import warnings
warnings.filterwarnings('ignore')

drive.mount('/content/drive')


ruta = '/content/drive/MyDrive/Ciencia de datos /dataset_limpio_transformado (2).xlsx'
df = pd.read_excel(ruta)
df['Fecha_DT'] = pd.to_datetime(df['Fecha_DT'])

# preparar lienzo de 4 cuadrantes (2x2)
fig, axs = plt.subplots(2, 2, figsize=(16, 10))
fig.suptitle('dashboard analitico integrador de salud', fontsize=18, fontweight='bold')

# superior izquierdo: regresion de peso y linea meta
sns.regplot(data=df, x=df.index, y='Peso_Kg', ax=axs[0, 0], scatter_kws={'alpha':0.5}, line_kws={'color':'orange'})
axs[0, 0].axhline(70.28, color='red', linestyle='--', label='meta (70.28 kg)')
axs[0, 0].set_title('tendencia de peso y regresion', fontweight='bold')
axs[0, 0].set_xlabel('dias de registro')
axs[0, 0].legend()

# superior derecho: dona del historico macro de 4 años
labels_dona = ['ciclismo exterior', 'ciclismo interior', 'caminatas', 'mancuernas', 'otros']
sizes_dona = [47, 25, 8, 3, 17] # los datos macro que pide la rubrica
axs[0, 1].pie(sizes_dona, labels=labels_dona, autopct='%1.1f%%', startangle=90,
              wedgeprops=dict(width=0.3), colors=sns.color_palette('pastel'))
axs[0, 1].set_title('distribucion historica de tiempo por deporte 4 años', fontweight='bold')

# 3. inferior izquierdo: dispersion (scatter plot) carga vs grasa
sns.regplot(data=df, x='carga_semanal_km', y='Grasa_%', ax=axs[1, 0], color='green')
axs[1, 0].set_title('carga semanal mtb vs porcentaje de grasa', fontweight='bold')
axs[1, 0].set_xlabel('km semanales (mtb)')

# inferior derecho: barras combinando picos de actividad
# filtramos solo los dias donde la smartband si registro actividad o distancia
df_actividad = df[(df['conteo_pasos'] > 0) | (df['Distancia_MTB_km'] > 0)].tail(15)

if not df_actividad.empty:
    sns.barplot(data=df_actividad, x='Fecha_DT', y='Distancia_MTB_km', ax=axs[1, 1], palette='Blues')
    axs[1, 1].set_title('picos de actividad recientes', fontweight='bold')
    axs[1, 1].tick_params(axis='x', rotation=45)
    axs[1, 1].set_xlabel('fecha')
    axs[1, 1].set_ylabel('kilometraje / actividad')
else:
    axs[1, 1].text(0.5, 0.5, 'no hay picos registrados en este corte', ha='center', va='center')

# ajustar espacios y mostrar grafico
plt.tight_layout()
plt.show()
```
